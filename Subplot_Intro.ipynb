```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
#fig = Canvas
# ax = single plot(Subpolot)
fig, ax = plt.subplots(2,2) #Rows x Column (2x2) gra
plt.show()
```

```python
x = np.array([1,2,3,4,5])
fig, ax = plt.subplots(2,2)
ax[0,0].plot(x, x*2 , color ='red')
ax[0,0].set_title("x*2")
plt.show()
```

```python
x = np.array([1,2,3,4,5])
fig, ax = plt.subplots(2,2)
ax[0,0].plot(x, x*2 , color ='red')
ax[0,0].set_title("x*2")
ax[0,1].plot(x, x**2 , color ='blue')
ax[0,1].set_title("x**2")
plt.show()
```

```python
x = np.array([1,2,3,4,5])
fig, ax = plt.subplots(2,2)
ax[0,0].plot(x, x*2 , color ='red')
ax[0,0].set_title("x*2")
ax[0,1].plot(x, x**2 , color ='blue')
ax[0,1].set_title("x**2")
ax[1,0].plot(x, x**3 , color ='blue')
ax[1,0].set_title("x**3")
plt.show()
```

```python
#Notice the Title is overlapping with the graph so we use plt.tight_layout()
```

```python
x = np.array([1,2,3,4,5])
fig, ax = plt.subplots(2,2)
ax[0,0].plot(x, x*2 , color ='red')
ax[0,0].set_title("x*2")
ax[0,1].plot(x, x**2 , color ='blue')
ax[0,1].set_title("x**2")
ax[1,0].plot(x, x**3 , color ='green')
ax[1,0].set_title("x**3")
ax[1,1].plot(x, x**4 , color ='orange')
ax[1,1].set_title("x**4")
plt.tight_layout()
plt.show()
```
